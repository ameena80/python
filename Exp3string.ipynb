```python
#program to change a given string to a new string
s=input("enter a new string:")
newstring=s[-1]+s[1:-1]+s[0]
print("new string:",newstring)
```

```python
# program to get a string(first 2 and last 2)
s=input("enter a string:")
newstring=s[:2]+s[-2:]
print("new string:",newstring)
```

```python
#program to swapping string
s1=input("enter a string:")
s2=input("enter next string:")
new=s2[0]+s1[1:]+""+s1[0]+s2[1:]
print("swap:",new)
```
