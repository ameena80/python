```python
#length of longest word
def longest(word):
    long=0
    for w in words:
        if len(w)>long:
            long=len(w)
    print("length of longest word:",long)
words = input("Enter a list of words: ").split()
longest(words)
```

```python
#nested loop
def pattern(n):
    for i in range(1,n+1):
        for j in range(i):
            print ("*",end="")
        print()
    for i in range(n-1,0,-1):
        for j in range(i):
            print("*",end="")
        print()
n=int(input("Enter the range:"))
pattern(n)
```

```python
#factors of a number
def fact(n):
    print("Factors are:")
    for i in range(1,n+1):
        if n%i==0:
             print(i)
n=int(input("Enter a number:"))
fact(n)
```

```python
# Area of a square
square = lambda a: a * a

# Area of a rectangle
rectangle = lambda length, breadth: length * breadth

# Area of a triangle
triangle = lambda base, height: 0.5 * base * height

a=float(input("enter side of square: "))
print("Area of square:",square(a))

l=float(input("enter length of rectangle:"))
b=float(input("enter breadth of rectangle:"))
print("Area of rectangle:", rectangle(l,b))

base=float(input("enter base of triangle:"))
height=float(input("enter length of triangle:"))
print("Area of triangle:", triangle(base,height))
```
