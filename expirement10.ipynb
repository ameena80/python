```python
#Count each word in a text
text=input("Enter the text:")
words=text.split()
print("text is:",text)
for w in set(words):
    print(w,":",words.count(w))
```

```python
#replace the value greater than 100 with "over"
n=list(map(int,input("Enter integers separated by spaces:").split()))
print("Numbers are:",n)
for i in range(len(n)):
    if n[i]>100:
        n[i]="over"
print("result is :",n)
```

```python
#Count the occurence of "a"
names=input("Enter first name:").split()
print("Names:",names)
count=0
for name in names:
    count +=name.lower().count('a')
print("Occurence of 'a':",count)
```

```python
#Find GCD of two Numbers
a=int(input("Enter first number:"))
b=int(input("Enter second number:"))
print("Enter the first number:",a)
print("Enter the second number:",b)
while b!=0:
    a,b=b,a%b
print("GCD=",a)
```

```python
#create a list by removing even numbers
numbers = list(map(int, input("Enter intergers separated by space:").split()))
result = []
for n in numbers:
    if n % 2 != 0:
        result.append(n)
print("list after removing even numbers:", result)        
```

```python
#create fibonacci series of N terms
n = int(input("Enter numbers of terms: "))
print("Enter number of terms:",n)
a = 0
b = 1
for i in range(n):
    print(a, end=" ")
    a, b = b, a + b
```

```python
#Four-digit numbers with all digits even and perfect square
start = int(input("Enter starting value: "))
end = int(input("Enter ending value: "))
print("Starting value=",start)
print("Ending value=",end)

import math

result = []

for n in range(start, end + 1):
    if 1000 <= n <= 9999:
        digits = str(n)

        if all(int(d) % 2 == 0 for d in digits):
            if math.isqrt(n) ** 2 == n:
                result.append(n)

print("Result:", result)
```

```python
#Count character frequency in a string
text = input("Enter a string: ")
print("Enter a string:",text)

frequency = {}

for char in text:
    frequency[char] = frequency.get(char, 0) + 1

print("Character frequency:", frequency)
```

```python
#Compare two lists
list1 = list(map(int, input("Enter first list: ").split()))
list2 = list(map(int, input("Enter second list: ").split()))
print("list one=",list1)
print("list two=",list2)

if len(list1) == len(list2):
    print("Lists have same length")
else:
    print("Lists do not have same length")

if sum(list1) == sum(list2):
    print("Lists have same sum")
else:
    print("Lists do not have same sum")

common = []

for n in list1:
    if n in list2:
        common.append(n)

if common:
    print("Common values:", common)
else:
    print("No common values")
```

```python
#Pyramid pattern
n = int(input("Enter N: "))
print("Enter number:",n)

for i in range(1, n + 1):
    for j in range(1, i + 1):
        print(i * j, end=" ")
    print()
```

```python
#find the factorial of a number
n = int(input("Enter a number: "))
print("number=",n)
factorial = 1
for i in range(1, n + 1):
    factorial = factorial * i
print("Factorial =", factorial) 
```
