```python
#bank account details
class BankAccount:
    def __init__(self, name, no, acc_type, balance):
        self.acc_name = name
        self.acc_no = no
        self.acc_type = acc_type
        self.acc_balance = balance

    def deposit(self, amount):
        print("Initial balance:", self.acc_balance)
        self.acc_balance += amount
        print(amount, "is deposited")
        print("Current balance:", self.acc_balance)

    def withdraw(self):
        amount = int(input("Withdraw amount: "))
        print("Initial balance:", self.acc_balance)

        if amount > self.acc_balance:
            print("You don't have enough balance to withdraw!")
            print("Current balance:", self.acc_balance)
        else:
            print(amount, "is withdrawn")
            self.acc_balance -= amount
            print("Current balance:", self.acc_balance)

    def acc_info(self):
        print("Account holder Name:", self.acc_name)
        print("Account Number:", self.acc_no)
        print("Account Type:", self.acc_type)
        print("Account Balance:", self.acc_balance)


name = input("Enter name: ")
no = int(input("Enter account number: "))
acc_type = input("Enter account type: ")
bal = int(input("Enter initial balance: "))

holder = BankAccount(name, no, acc_type, bal)

while True:
    print("\n1. Deposit 2. Withdraw 3. Account info 4. Exit")
    ch = int(input("Choose your option: "))

    if ch == 1:
        amount = int(input("Deposit amount: "))
        holder.deposit(amount)
    elif ch == 2:
        holder.withdraw()
    elif ch == 3:
        holder.acc_info()
    elif ch == 4:
        print("Exiting...")
        break
    else:
        print("Invalid choice")
```

```python
#book details
class Publisher:
    def __init__(self, pname):
        self.pubname = pname

    def display1(self):
        print("Book details")
        print("Publisher name:", self.pubname)


class Book(Publisher):
    def __init__(self, pname, bname, author):
        self.title = bname
        self.auth = author
        Publisher.__init__(self, pname)

    def display2(self):
        print("Title:", self.title)
        print("Author:", self.auth)


class Python(Book):
    def __init__(self, pname, title, author, price, pages):
        self.rate = price
        self.nop = pages
        Book.__init__(self, pname, title, author)

    def display3(self):
        print("Rate:", self.rate)
        print("No. of pages:", self.nop)


p = input("Enter publisher name: ")
b = input("Enter book name: ")
a = input("Enter author name: ")
r = int(input("Enter rate of book: "))
n = int(input("Enter no. of pages: "))

obj = Python(p, b, a, r, n)

obj.display1()
obj.display2()
obj.display3()
```

```python
#Area of two rectangle class
class Rectangle:
    def __init__(self, l, w):
        self.__length = l
        self.__width = w
        self.__area = l * w

    def __lt__(self, other):
        if self.__area < other.__area:
            print("Rectangle 1 is smaller")
        else:
            print("Rectangle 2 is smaller")


l1 = int(input("Enter length of rectangle 1: "))
w1 = int(input("Enter width of rectangle 1: "))

l2 = int(input("Enter length of rectangle 2: "))
w2 = int(input("Enter width of rectangle 2: "))

r1 = Rectangle(l1, w1)
r2 = Rectangle(l2, w2)

r1 < r2
```

```python
#time class
class Time:
    def __init__(self, h, m, s):
        self.__hour = h
        self.__minute = m
        self.__second = s

    def __add__(self, other):
        total = (self.__hour * 3600 +
                 self.__minute * 60 +
                 self.__second)

        total += (other.__hour * 3600 +
                  other.__minute * 60 +
                  other.__second)

        h = total // 3600
        m = (total % 3600) // 60
        s = total % 60

        return Time(h, m, s)

    def display(self):
        print("Time:", self.__hour, "hours",
              self.__minute, "minutes",
              self.__second, "seconds")


h1 = int(input("Enter hour 1: "))
m1 = int(input("Enter minute 1: "))
s1 = int(input("Enter second 1: "))

h2 = int(input("Enter hour 2: "))
m2 = int(input("Enter minute 2: "))
s2 = int(input("Enter second 2: "))

t1 = Time(h1, m1, s1)
t2 = Time(h2, m2, s2)

t3 = t1 + t2

print("Sum of two times:")
t3.display()
```
