```python
#positive list of numbers
n=list(map(int,input("enter the numbers=").split(",")))
print("numbers=",n)
positive=[x for x in n if x>0]
print("positive=",positive)
```

```python
#squeres of N numbers
n=int(input("Enter limit="))
print("limit=",n)
squeres=[x**2 for x in range(1,n+1)]
print("squers=",squeres)
```

```python
#Form a list of vowels 
w=input("Enter a word:")
print("Word=",w)
vowels=[ch for ch in w if ch in "aeiou"]
print("Vowels=",vowels)
```

```python
#list ordinal value of each element
w=input("Enter a word=")
print("word=",w)
ordinal=[ord(ch) for ch in w]
print("ordinal values=",ordinal)
```
