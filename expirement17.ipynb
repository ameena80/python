```python

%%writefile input.txt
red
blue
black
green
purple
brown
```

```python
#Copy Odd Lines to a New File
with open("input.txt","r") as src:
    lines=src.readlines()
with open("odd_lines.txt","w") as dest:
    for i in range(len(lines)):
        if i % 2 ==0:
            dest.write(lines[i])
print("Odd lines copied successfully!")
```

```python
#Write a Python dictionary into a CSV file
import csv
studentdetail=[
    {'Name':'John','Age':21,'Place':'Calicut'},
    {'Name':'Ann','Age':20,'Place':'Palakkad'},
    {'Name':'Riya','Age':23,'Place':'Kannur'}]
with open("studentdetail.csv","w",newline="") as file:
    fields=['Name','Age','Place']
    writer=csv.DictWriter(file,fieldnames=fields)
    writer.writeheader()
    writer.writerows(studentdetail)
print("Dictionary written successfully!")
```
