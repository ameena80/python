```python
#print colors in set1 not in set2
s1=set(input("Enter colors for set1(separated by commas):").split(","))
s2=set(input("Enter colors for set2(separated by commas):").split(","))
print("set1:",s1)
print("set2:",s2)
print("colors in set1 not in set2:",s1-s2)
```

```python
#print colors in set1 and set2
s1=set(input("Enter colors for set1(separated by commas):").split(","))
s2=set(input("Enter colors for set29separated by commas:").split(","))
print("set1:",s1)
print("set2:",s2)
print("colors in set1 and set2:",s1|s2)
```
