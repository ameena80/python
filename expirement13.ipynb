```python
def area(l,b):
    return l*b
def perimeter(l,b):
    return 2*(l+b)
```

```python
import math

def area(r):
    return math.pi*r*r

def perimeter(r):
    return 2 * math.pi * r
```

```python
import Rectangle
import Circle

while True:
    print("1.Rectangle")
    print("2.Circle")
    print("3.Exit")

    ch=int(input("Enter your choice:"))

    if ch==1:
        l=int(input("Enter length:"))
        b=int(input("Enter breadth:"))

        print("area of Rectangle:",Rectangle.area(l,b))
        print("Perimeter of Rectangle:",Rectangle.perimeter(l,b))

    elif ch==2:
          r=int(input("Enter radius:"))
          print("Area of Circle;",round(Circle.area(r),2))
          print("perimeter of Circle:",round(Circle.perimeter(r),2))

    elif ch ==3:
          break

    else:
          print("Invalid choice")
```
