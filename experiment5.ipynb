```python
#python program to get third element
n=input("enter the numbers:")
t=tuple(map(int,n.split(",")))
print("tuple:",t)
print("Third element:",t[2])
print("Third element from last:",t[-3])
```

```python
#python program to comma-seperated numbers
N=input("enter comma-seperated numbers:")
t=tuple(map(int, N.split(",")))
print("tuple:",t)
print("avarage:",sum(t)/len(t))
```
